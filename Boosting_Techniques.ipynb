# **THEORETICAL QUESTIONS**

## Q1. What is Boosting in Machine Learning?

## ANS 1.

- Boosting is a machine learning method that combines many simple models (called weak learners) to make one strong model.
- It trains models sequentially, where each new model focuses on correcting the errors of the previous ones to improve accuracy.



## Q2. How does Boosting differ from Bagging?

## ANS 2.

Following are the differences between Boosting and Bagging:

**Boosting:**

1. Models are built sequentially - each learns from the previous one's errors.
2. Focuses on reducing bias.
3. Gives more weight to wrong predictions to improve them in the next round.
4. Models are dependent on each other.
5. Higher risk of overfitting.

**Bagging:**

1. Models are built in parallel - independently from each other.
2. Focuses on reducing variance.
3. Uses random subsets of data (with replacement).
4. Models are independent of each other.
5. Less risk of overfitting.


## Q3. What is the key idea behind AdaBoost?

## ANS 3.

The key idea behind AdaBoost is described as follows:

1. Start simple: Train a weak model (e.g., a small decision tree).
2. Check errors: Identify which data points were misclassified.
3. Update weights: Increase the importance (weight) of misclassified points.
4. Train next model: Focus more on these hard cases.
5. Repeat: Combine all models using weighted votes.

## Q4. Explain the working of AdaBoost with an example.

## ANS 4.

#### **Working of AdaBoost:**

1. Start with equal weights for all training samples.
2. Train a weak learner (e.g., a small decision tree).
3. Find the misclassified points.
4. Increase weights of the misclassified points.
5. **Train next model**: Focus more on the difficult (high-weight) samples.
6. Repeat for a fixed number of rounds.
7. **Final prediction**: Combine all models using a weighted vote.


#### **Example: AdaBoost (Binary Classification)**

Suppose we have a dataset:

Data Point : A, B, C, D  
Feature    : 1, 2, 3, 4  
Label : +1, +1, -1, -1  
Weights : 0.25,0.25,0.25,0.25

#### Step 1: Initialize Weights

* All points get equal weight: **1/4 = 0.25**

#### Step 2: Train Weak Learner #1

* It misclassifies point **C**.

#### Step 3: Update Weights

* Increase weight for **C** (hard example).
* Decrease weights for correctly classified points.

#### Step 4: Train Weak Learner #2

* Focuses more on **C**. This time it correctly classifies **C**, but misclassifies **B**.

#### Step 5: Update Again

* Increase weight for **B** (now hard).

#### Step 6: Repeat this process...

#### Final Prediction:

* Combine all weak learners, giving more weight to the more accurate ones.
* Final output = weighted majority vote of all weak learners.


## Q5. What is Gradient Boosting, and how is it different from AdaBoost?

## ANS 5.

- Gradient Boosting is a boosting technique that builds models sequentially, where each new model tries to correct the errors (residuals) of the previous models using gradient descent.

- Difference between Gradient Boosting and AdaBoost:  
Instead of adjusting sample weights like AdaBoost, Gradient Boosting fits new models to the residual errors (difference between actual and predicted values) of the previous model — like correcting the leftover mistakes using gradients.



## Q6. What is the loss function in Gradient Boosting?

## ANS 6.

- The **loss function** in Gradient Boosting measures how far the model's predictions are from the actual values.
- It tells the model how wrong it is.
- Gradient Boosting uses the gradient (slope) of this loss to decide how to correct the error in the next step.

## Q7. How does XGBoost improve over traditional Gradient Boosting?

## ANS 7.

XGBoost Improves Over Gradient Boosting by the following means:

1. **Faster Training**: Uses optimized algorithms.
2. **Regularization**: Adds L1 & L2 regularization to prevent overfitting.
3. **Missing Data Handling**: Automatically handles missing values.
4. **Scalability**: Can handle large datasets more efficiently.
5. **Built-in Cross-Validation**: Simplifies model tuning.


## Q8. What is the difference between XGBoost and CatBoost?

## ANS 8.

 Following are the differences between XGBoost and CatBoost:

#### **XGBoost:**

1. Requires manual encoding of categorical features (e.g., one-hot).
2. Very fast training, especially on numerical data.
3. Uses regularization (L1 & L2) to reduce overfitting.
4. Works best with numerical or already encoded data.

#### **CatBoost:**

1. Automatically handles categorical features - no need for manual encoding.
2. Slightly slower training, but faster predictions.
3. Uses ordered boosting to prevent overfitting.
4. Easier to use with mixed data (numerical + categorical).

## Q11. What are some hyperparameters to tune in Gradient Boosting models?

## ANS 11.

Key hyperparameters to tune in Gradient Boosting models are:

1. **`n_estimators`**
   * Number of boosting rounds (trees).
   * Too high → overfitting, too low → underfitting.

2. **`learning_rate`**
   * Lower = better generalization, but needs more trees.

3. **`max_depth`**
   * Maximum depth of each tree.

4. **`subsample`**
   * Fraction of training data used per tree.
   * Helps prevent overfitting (like dropout).

Tuning hyperparameters like `learning_rate`, `n_estimators`, and `max_depth` helps balance **bias vs. variance** and improve **model accuracy**.


## Q12. What is the concept of Feature Importance in Boosting?

## ANS 12.

**Feature Importance in Boosting:**

- Shows how much each feature contributes to model predictions.
- Measured by how often and how effectively a feature is used in tree splits.
- Helps with feature selection, model understanding, and interpretability.
- More important features reduce the model's loss/error more when used.
- Example:  
  - If feature A is used in most trees and splits the data well, it will have high importance.
  - If feature B is rarely used or doesn't reduce error, it will have low importance.




## Q13. Why is CatBoost efficient for categorical data?

## ANS 13.

CatBoost is efficient for categorical data because of the following reasons:

1. No manual encoding needed - handles categories automatically.
2. Uses ordered target statistics to avoid target leakage.
3. Works well with high-cardinality features (many unique values).
4. Automatically creates feature combinations for better learning.
5. Reduces overfitting and improves accuracy on mixed data.

## Q18. Train an XGBoost Classifier on a dataset and compare accuracy with Gradient Boosting.

## Q22. Train a Gradient Boosting Regressor and plot learning curves.

```python
x22, y22 = make_regression(n_samples=1000, n_features=10, noise=10, random_state=1)

x22_train, x22_test, y22_train, y22_test = train_test_split(x22, y22, test_size=0.2, random_state=1)

gb_reg22 = GradientBoostingRegressor()
gb_reg22.fit(x22_train, y22_train)
y22_pred = gb_reg22.predict(x22_test)

from sklearn.model_selection import learning_curve

train_sizes, train_scores, test_scores = learning_curve(
    gb_reg22, x22, y22, scoring='neg_mean_squared_error',
    train_sizes=np.linspace(0.1, 1.0, 10), cv=5, n_jobs=-1)

train_errors = -np.mean(train_scores, axis=1)
test_errors = -np.mean(test_scores, axis=1)

plt.figure(figsize=(8, 6))
plt.plot(train_sizes, train_errors, 'o-', color='blue', label='Training Error')
plt.plot(train_sizes, test_errors, 'o-', color='green', label='Validation Error')
plt.title('Learning Curves - Gradient Boosting Regressor')
plt.xlabel('Training Set Size')
plt.ylabel('Mean Squared Error')
plt.legend()
plt.grid(True)
plt.show()
```

## Q25. Train an AdaBoost Classifier with different numbers of estimators and compare accuracy.

```python
x25, y25 = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

x25_train, x25_test, y25_train, y25_test = train_test_split(x25, y25, test_size=0.3, random_state=1)

accuracies25 = []

for i in [1, 10, 20, 40, 50, 100, 200, 250, 300, 500]:
  ada_clf25 = AdaBoostClassifier(n_estimators=i)
  ada_clf25.fit(x25_train, y25_train)
  y25_pred = ada_clf25.predict(x25_test)

  acc25 = accuracy_score(y25_test, y25_pred)
  print(f"Accuracy for estimator = {i}: {acc25: .4f}")
  accuracies25.append(acc25)

best_accuracy = max(accuracies25)
best_index = accuracies25.index(best_accuracy)
best_n_estimator = [1, 10, 20, 40, 50, 100, 200, 250, 300, 500][best_index]
print(f"Best n_estimator: {best_n_estimator}")
```

## Q28. Train a CatBoost Classifier on an imbalanced dataset and compare performance with class weighting.

```python
x28, y28 = make_classification(n_samples=1000, n_features=20, n_informative=15,
                           n_redundant=5, weights=[0.1, 0.9], random_state=1)

x28_train, x28_test, y28_train, y28_test = train_test_split(x28, y28, test_size=0.3, random_state=1)

#Without Weights
model_no_weights = CatBoostClassifier(verbose=0, random_state=1)
model_no_weights.fit(x28_train, y28_train)
y_pred_no_weights = model_no_weights.predict(x28_test)

print("Performance WITHOUT Class Weights:")
print(classification_report(y28_test, y_pred_no_weights))

#With Weights
# Here class 0 is minority, class 1 is majority
# Set higher weight for minority class
model_with_weights = CatBoostClassifier(
    class_weights=[10, 1],  # [weight_for_class_0, weight_for_class_1]
    verbose=0,
    random_state=1)
model_with_weights.fit(x28_train, y28_train)
y_pred_with_weights = model_with_weights.predict(x28_test)

print("Performance WITH Class Weights:")
print(classification_report(y28_test, y_pred_with_weights))
```

## Q29. Train an AdaBoost Classifier and analyze the effect of different learning rates.

```python
x29, y29 = make_classification(n_samples=1000, n_features=10, n_classes=2, random_state=1)

x29_train, x29_test, y29_train, y29_test = train_test_split(x29, y29, test_size=0.3, random_state=1)

learning_rates29 = [0.01, 0.1, 0.5, 1.0, 2.0]
train_scores29 = []
test_scores29 = []

for rate in learning_rates29:
    model29 = AdaBoostClassifier(n_estimators=100, learning_rate=rate, random_state=1)
    model29.fit(x29_train, y29_train)

    train_acc = accuracy_score(y29_train, model29.predict(x29_train))
    test_acc = accuracy_score(y29_test, model29.predict(x29_test))

    train_scores29.append(train_acc)
    test_scores29.append(test_acc)

    print(f"Learning Rate: {rate}")
    print(f"Train Accuracy: {train_acc:.4f} | Test Accuracy: {test_acc:.4f}\n")
```
